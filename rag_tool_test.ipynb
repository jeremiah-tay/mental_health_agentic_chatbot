```python
import os
import pandas as pd
import json
from backend.agentools.rag_tool import rag_tool
```

```python
# --- 1. Run RAG Tool Query
query = "What are the symptoms of social anxiety disorder?"
result = rag_tool.invoke(query)

answer = result["answer"]
docs = result["retrieved_docs"]

print("Answer:\n", answer)
print("\nRetrieved chunks:\n")
for i, d in enumerate(docs, 1):
    print(f"{i}. {d['content'][:300]}... (source: {d['source']})")
```

```python
# --- 2. Save outputs
os.makedirs("outputs", exist_ok=True)

# Save retrieved docs to CSV
df = pd.DataFrame(docs)
df.to_csv("outputs/rag_tool_results.csv", index=False)

# Save log with query, answer, and docs
with open("outputs/rag_tool_log.json", "w") as f:
    json.dump(
        {"query": query, "answer": answer, "retrieved": docs},
        f,
        indent=2
    )

print("\n Saved outputs to outputs")
```
